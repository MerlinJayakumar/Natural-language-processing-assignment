```python
import pandas as pd

df = pd.read_csv(
    r"D:\HOPE\AI COURSE TAMIL\NATURAL LANGUAGE PROCESSING\Whatsapp demo\Assignment 5\missed _call.txt",
    header=None,
    names=['Person','chat']
)

print(df.head())
```

```python
missed_calls = df[
    df['chat'].str.contains('Missed',case=False,na=False)
]

print(missed_calls['Person'].value_counts())
```

```python
missed_calls['Person'].value_counts().plot(kind='bar')

plt.title('Missed Calls by Person')
plt.xlabel('Person')
plt.ylabel('Missed Calls')
plt.show()
```
