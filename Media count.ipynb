```python
import pandas as pd
```

```python
df=pd.read_csv('media_count.txt')
```

```python
print(df.head())
print(df.shape)
```

```python
media_count = df.groupby('Sender').size()
```

```python
print("Media Counts Sent by Each Person:")
print(media_count)
```

```python
import matplotlib.pyplot as plt

media_count.plot(kind='bar')

plt.title('Media Counts Sent by Each Person')
plt.xlabel('Person')
plt.ylabel('Number of Media Sent')

plt.show()
```
