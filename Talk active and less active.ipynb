```python
import pandas as pd
import re
```

```python
file=open("mabel.txt","r",encoding='utf-8')
```

```python
lines=file.readlines()
```

```python
data = []

for line in lines:
    match = re.match(r'^(\d{1,2}/\d{1,2}/\d{2}),\s*(\d{1,2}:\d{2}\s*[ap]m)\s-\s([^:]+):\s(.*)', line.strip(), re.I)

    if match:
        sender = match.group(3)
        chat = match.group(4)

        data.append([sender, chat])
```

```python
df=pd.DataFrame(data,columns=['Sender','Chat'])

print(df.head())
```

```python
activity = df.groupby('Sender')['Chat'].count().reset_index()

activity.columns = ['Sender', 'Message_Count']

print(activity)
```

```python
activity['Activity'] = activity['Message_Count'].apply(
    lambda x: 'Talkative' if x > activity['Message_Count'].median()
    else 'Less Talkative'
)
```
