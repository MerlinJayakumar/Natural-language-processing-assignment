```python
import pandas as pd
```

```python
dataset=pd.read_csv('time_encounters.txt')
```

```python
print(dataset.head())
print(dataset.shape)
```

```python
dataset.info()
dataset.isnull().sum()
dataset.columns
```

```python
dataset['Date'] = pd.to_datetime(
    dataset['Date'], dayfirst=True
)
```

```python
date_count = dataset['Date'].dt.date.value_counts()

print(date_count)
print("Most Active Date:")
print(date_count.idxmax())
```

```python
dataset['Day'] = dataset['Date'].dt.day_name()

day_count = dataset['Day'].value_counts()

print(day_count)
print("Most Active Day:")
print(day_count.idxmax())
```

```python
import matplotlib.pyplot as plt

day_order = ['Monday', 'Tuesday', 'Wednesday',
             'Thursday', 'Friday', 'Saturday', 'Sunday']

day_count = day_count.reindex(day_order, fill_value=0)

day_count.plot(kind='bar')

plt.title('Most Active Day')
plt.xlabel('Day')
plt.ylabel('Number of Messages')
plt.show()
```
